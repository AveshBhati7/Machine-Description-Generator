```python
import pandas as pd

# Load the CSV file
file_path = '/Users/arthedain/Desktop/Avesh/Machine Description/Code/Machine_Description_Complete.csv'
df = pd.read_csv(file_path)
```

```python
df.rename(columns={'Machine Id': 'MachineId','About':'AboutUsDescription'}, inplace=True)
```

```python
# Filter to keep only 'machine id' and 'about' columns
df_filtered = df[['MachineId','AboutUsDescription']]
```

```python
records = df_filtered.to_dict(orient='records')
```

```python
api_url=''
api=''
```

```python
import requests

api_url = api_url
headers = {
    'accept': 'text/plain',
    'ApiKey': api,
    'Content-Type': 'application/json'
}

for index, row in df_filtered.iterrows():
    data = {
        "MachineId": row['MachineId'],
        "AboutUsDescription": row['AboutUsDescription']
    }

    response = requests.post(api_url, headers=headers, json=data)

    if response.status_code == 200:
        print(f"Successfully inserted: {data}")
    else:
        print(f"Failed to insert: {data}, Status Code: {response.status_code}, Response: {response.text}")
```
